```python
#CLEANING DATA

#Removing cells which contain empty cells using dropna() function
import pandas as pd
df= pd.read_csv('C:/Users/Sham/Desktop/Healthcare.csv')
new_df=df.dropna()
print("1. dropna() function to remove the cells which contain empty cells:")
print(new_df.to_string(), ("\n"))
#Does'nt change the original dataset. To make changes in the original dataset use; print(df.dropna(inplace = True))


#Replace empty values using fillna() function
import pandas as pd
df= pd.read_csv('C:/Users/Sham/Desktop/Healthcare.csv')
print ("2. fillna() function to replace the cells which contain empty cells:")
df.fillna(130,inplace = True)
print(df.to_string(), ("\n"))


#Replace only for specified columns
import pandas as pd
df = pd.read_csv('C:/Users/Sham/Desktop/Healthcare.csv')
print ("3. Replace the empty cells of only the specified column:")
df.fillna({"Calories": 130}, inplace=True)
print(df.to_string(), ("\n"))


#Replace using mean, median or mode
import pandas as pd
df = pd.read_csv('C:/Users/Sham/Desktop/Healthcare.csv')
x = df["Calories"].mean()
print ("4. Replace the empty cells using mean, median or mode:")
df.fillna({"Calories": x}, inplace=True)
print (df.to_string(), ("\n"))


#CLEANING WRONG DATA

#Replacing values
import pandas as pd
df.loc[7, 'Duration']=450
print ("5. Replacing values of the 7th duration:")
print(df.to_string(), ("\n"))

#Removing rows
import pandas as pd
for x in df.index:
    if df.loc[x,"Duration"]>120:
        df.drop(x,inplace=True)
print ("6. Removing rows which have value greater than 120 (i.e. 7th row):")
print (df.to_string(), ("\n"))

#Discovering Duplicates
import pandas as pd
print("7. Discovering Duplicate values:")
print(df.duplicated(), ("\n"))

#Removing Duplicate values
import pandas as pd
print("8. Removing Duplicate values:")
df.drop_duplicates(inplace=True)
print (df.to_string(), ("\n"))

#Finding Relationships 
#corr() method calculates the relationship between each column in the dataset
import pandas as pd
print("9. Finding relationships:")
print (df.corr(), ("\n"))
```
