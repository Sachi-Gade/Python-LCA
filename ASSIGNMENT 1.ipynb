```python
#Create lits
mylist= ["apple", "banana", "pineapple", "mango", "kiwi", "chiku"]
print ("Original list:", (mylist))

#Length of the list
print ("1. Output of length of the list:",len(mylist))

#Access item from list
print ("2. Output of access elements:", (mylist[0]))

#Negative Indexing
print ("3. Output of negative indexing:", (mylist[-3]))

#Range of indexes
print ("4. Output of range of indexes:", (mylist[2:5]))

#Range of negative indexing
print ("5. Output of range of negative indexing:", (mylist[-2:]))

#Check if item exists
if "pineapple" in mylist:
    print ("6. Output of checking if item exists:", "Yes, pineapple exists in the list")

#Change item from list
mylist [1]= "orange"
print ("7. Output of change item:", (mylist))

#Change a range of items:
mylist [1:3]= ["blackcurrent", "watermelon"]
print ("8. Output of change a range of items:", (mylist))

#Insert items
mylist.insert (2, "cherry")
print ("9. Output of insert item:", (mylist))

#Add item:
mylist.append ("muskmelon")
print ("10. Output of add item:", (mylist))

#Extend list: append elements from another list to the current list (*not necessary that it should be a list only,it can also be a tuple)
seasons= ["summer", "winter", "autumn", "spring", "monsoon"]
mylist.extend(seasons)
print ("11. Output of extend list:", (mylist))

#Remove item from list (removes only one element)
mylist.remove("muskmelon")
print ("12. Output of remove element:", (mylist))

#Remove specified index (if index not specified removes last element)
mylist.pop(3)
print ("13. Output of remove specified index:", (mylist))

#Remove using del keyword:
del mylist[1]
print ("14. Output of del keyword:", (mylist))

#Sort item in list (in ascending order)
mylist.sort()
print ("15. Output of sort list in ascending order:", (mylist))

#Sort list in descending order
mylist.sort(reverse=True)
print("16. Output of sort list in descending order:",(mylist))

#Reversing the list
mylist.reverse()
print ("17. Output of reversing the list:", (mylist))

#Copy the list
a= mylist.copy()
print ("18. Output of copy the list:", (a))

#Slicing the list using slice (:) operator
b= mylist[2:8]
print ("19. Output of slicing the list:", (b))

#Joining two list
list1= ["1", "2", "3", "4", "5"]
c= mylist + list1
print ("20. Output of joining lists:", (c))

#Counting number of mentioned elements in the list (:.1 as cherry is once in the list)
d= mylist.count ("cherry")
print ("21. Output of count method:", (d))

#Clear the list
mylist.clear()
print ("22. Output of clear list:", (mylist))
```

```python
# Question: University Tech Fest Participant Management (LISTS)
students= ["Aarav","Isha","Rohan","Sneha","Kabir"]

#Access third student's name
print ("1. Output of accessing third student's name:", (students[2]))

#Change in spelling
students[2]= ("Rohan Patil")
print ("2. Output of changing the name:", (students))

#Add new students name to the list
newnames= ["Ananya", "Vihaan"]
students.extend(newnames)
print("3. Output of adding new names:", (students))

#Remove name from the list
students.pop(4)
print ("4. Output of removing the name:", (students))

#Total number of students in the list
print ("5. Output of total number of registered students:", (len(students)))

#Checking whether name Isha is in the list or not
if "Isha" in students:
    print("6. Output of checking for Isha's name in the list:", "Yes, Isha's name is in the list")

#Arranging list in alphabetical order
students.sort()
print ("7. Output of arranging list in alphabetical order:", (students))

#Creating copy of the list
backup_students= students.copy()
print("8. Output of making a copy of the list:", (backup_students))

#Combining lists
hackathon=["Meera","Arjun","Isha"]
all_participants= students + hackathon
print ("9. Output of combining lists:", (all_participants))

#Final display
print ("10. Final course display:", (students))
print (backup_students)
print (all_participants)
```

```python
#Create Tuple
mytuple= ("rose", "jasmine", "lotus", "marigold", "mogra")
print ("Original tuple:", (mytuple))

#Access item in tuple
print("1. Output of access item:", (mytuple[3]))

#Negative indexing in tuple
print ("2. Output of negative indexing:", (mytuple[-4]))

#Range of indexes in tuple
print ("3. Output of range of indexes:", (mytuple[2:4]))

#Range of negative indexes
print ("4. Output of range of negative indexes:", (mytuple[:-3]))

#Checking if item exists
if "lotus" in mytuple:
     print ("5. Output of checking if item exists:", "Yes, lotus exists in the tuple")

#Add item
y= list(mytuple)
y.append("lilies")
print ("6. Output of add item:", (y))

#Remove item
z= list(mytuple)
z.remove("marigold")
print ("7. Output of remove item:", (z))

#Unpacking the tuple
(rose, jasmine, lotus, marigold, mogra)= mytuple
print ("8. Output of unpacking the tuple:" )
print (rose)       
print (jasmine)
print (lotus)
print (marigold)
print (mogra)

#Joining tuples
tuple1= ("petals", "leaves", "branches", "bud")
a= mytuple + tuple1
print ("9. Output of joining tuple:", (a))

#Multiple tuples
flowers= mytuple * 2
print ("10. Output of multiplying tuple:", (flowers))

#Count in tuple
b= flowers.count(lotus)
print ("11. Output of counting the element in tuple:", (b))

#Index method in tuple (Index number of specified element is displayed)
a= mytuple.index(marigold)
print ("12. Output of index in tuple:", (a))
```

```python
# Maintaining the names of courses selected by a student (TUPLE):
courses= ("Python","Data Science","Cloud Computing","Cyber Security","AI")

#Displaying the second course selected by student
print ("1. Output of second course selected:", (courses[1]))

#Displying last two courses
print ("2. Output of displaying last two courses:", (courses[3:5]))

#Check whether Python present in the tuple
if "Python" in courses:
    print("3. Output for checking if python is there in the course:","Yes, Python is present in the course")

#Finding index of cloud computing from the tuple
a= courses.index("Cloud Computing")
print("4. Output of finding index of Cloud Computing:", (a))

#Total number of courses
print ("5. Output of total number of courses selected:", (len(courses)))

#Adding new courses in the existing tuple
new_enroll= ("Machine Learning", "Web Development")
updated = courses + new_enroll
print ("6. Output of adding new courses:", (updated))

#Removing cyber security from the tuple
c= list(updated)
c.remove("Cyber Security")
print ("7. Output of removing cyber security:", (c))

#Arrange in ascending
sorted_courses= tuple(sorted(updated))
print("8. Output of sorting the tuple:", (sorted_courses))

#Creating a copy of sorted_courses
backup_courses= sorted_courses
print("9. Output of copy of sorted_courses:", (backup_courses))

#Combining tuples
friend_courses= ("AI","Blockchian","Python")
combined_courses= sorted_courses + friend_courses
print ("10. Output of combining the tuples:", (combined_courses))

#Final display
print ("11. Final course display:", (updated))
print (sorted_courses)
print (backup_courses)
print (combined_courses)
```

```python
#Dictionary
mydict= {
    "brand": "ford",
    "model": "mustang",
    "year": "1964",
    "colors": ["red", "black", "white"]
}
print ("Original Dictionary:", (mydict))

#Length of the dictionary
print ("1. Output of length of dictionary:", (len(mydict)))

#Access item
a= mydict["model"]
print ("2. Output of access item:", (a))

#Access item using get method
b= mydict.get("brand")
print ("3. Output of access item using get command:", (b))

#Access keys of dictionary
c= mydict.keys()
print ("4. Output of accessing keys:", (c))

#Accessing values of dictionary
d= mydict.values()
print ("5. Output of accessing values:", (d))

#Get items (returns each item in dictionary)
e= mydict.items()
print ("6. Output of get items:", (e))

#Check if key exists
if "year" in mydict:
    print ("7. Output of check if key exists:", "Yes, 'year' is one of the keys in the dictionary")

#Change values
mydict["year"]= 2018
print ("8. Output of change values:", (mydict))

#Change using update method
mydict.update({"model": "endeavour"})
print ("9. Output of changes using update method:", (mydict))

#Adding items
mydict["tyres"]= "MRF"
print ("10. Output of adding items:", (mydict))

#Adding items using update method
mydict.update ({"interior": "beige"})
print ("11. Output of adding item using update method:", (mydict))

#Removing items
mydict.pop("brand")
print ("12. Output of removing item:", (mydict))

#Removing item using pop item method (removes last inserted item)
mydict.popitem()
print ("13. Output of removing item using pop item method:", (mydict))

#Removing item using del keyword (can also delete the complete dictionary)
del mydict ["colors"]
print ("14. Output of removing item using del keyword:", (mydict))

#Copying dictionary
mydict.copy()
print ("15. Output of copying dictionary:", (mydict))

# Copy of dictionary using dict() function
Dictionary= dict(mydict)
print ("16. Output of copy dictionary using dict() function:", (Dictionary))

# Using the clear() method (empties the dictionary)
mydict.clear()
print ("17. Output of using clear() method:", (mydict))

# Nested Dictionaries
classX = {
    "Emily": {
        "division": "A",
        "subjects": ("english", "math", "social science", "science"),
        "year": 2008
    },
    "Andrew": {
        "divison": "B",
        "subjects": ("english", "math", "science", "german"),
        "year": 2007
    },
    "Hannah": {
        "division": "B",
        "subjects": ("english", "french", "science"),
        "year": 2008
    }
}
print ("18. Output of nested dictionary:", (classX))

#Acces item in nested dictionary
print ("19. Output of accessing item in nested dictionary:", classX["Andrew"]["year"])

    
```
