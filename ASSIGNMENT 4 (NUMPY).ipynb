```python
#Addition of two arrays using numpy
import numpy as np
arr1= np.array([[1,2],[3,4]])
arr2=np.array ([[5,6],[7,8]])
sum= arr1+arr2
print ("Addition of two arrays using numpy:")
print (sum, ("\n"))


#Addition of two arrays using list
arr1=[[1,2],[3,4]]
arr2=[[5,6],[7,8]]
result=[[0,0], [0,0]]
for i in range(len(arr1)):
    for j in range (len(arr2)):
        result[i][j]= arr1[i][j] + arr2[i][j]
print ("Addition of two arrays using list:", result)
```
