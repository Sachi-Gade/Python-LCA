```python
import re
def check_string(text):
    pattern= r'^[a-zA-Z0-9]+$'
    if re.match(pattern,text):
        return "The string contains only valid characters (a-z, A-Z, 0-9)"
    else:
        return "The string contains invalid characters"
user_input= input ("Enter a string to check:")
print (check_string(user_input))
```

```python
import re

text = "The rain in Spain"
x = re.search("^The.*Spain$", text)
            # ^ = Starts with, . = any character except newline, * = zero or more occurrences, $ = Ends with

if x:
    print ("Yes! We have a match.")
else:
    print ("No match found")
```

```python
#Regex Functions

#1.findall()
import re
text= "The rain in Spain"
x=re.findall("ai",text)
print ("1. Output of findall() function:", x)

#2. search()
import re
text= "The rain in Spain"
x=re.search("\s", text)
print ("2. Output of search() function-{The first white space character is located in posoition}:", x.start())

#3. split()
import re
text= "The rain in the Spain"
x= re.split("\s", text)
print ("3. Output of split() function:", x)

#Controling the number of occurrences using maxsplit parameter
import re
text= "The rain in the Spain"
x=re.split("\s", text, 2)
print ("4. Output of maxsplit parameter:", x)

#4. sub() 
import re
text= "The rain in the Spain"
x=re.sub("\s", "9", text)    #Substituting blank spaces with 9
print ("5. Output of sub() function:", x)

#Controling the number of substitutions(replacements) using count() function
import re
text= "The rain in the Spain"
x=re.sub("\s", "9", text, 2)
print ("6. Output of count() parameter:", x)

#Match Object
import re
text= "The rain in the Spain"
x=re.search("ai", text)
print ("7. Output of match object:", x.start())

#
```
